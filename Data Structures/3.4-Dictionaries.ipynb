```python
## Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable(e.g strings, numbers, or tuples) while values can of any type.
```

```python
## Creating Dictionary 
empty_dict ={}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"Name": "Tarikul","age": 25, "grade": 2.5 }
print(student)
print(type(student))
```

```python
# Single key is always used 
student = {"name":"Tarikul", "age":25, "name":2.5}
print(student)
```

```python
## accessing dictionary elements 
student = {"name": "Tarikul", "age": 25, "grade": 'A'}
print(student)
```

```python
## Accessing dictionary elements 
print(student['grade'])
print(student['age'])


## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not available"))
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student["age"] = 33 ## Update value for the key
print(student)
student["address"]="Bangladesh" ## Added a new key and value
print(student)
```

```python
del student['grade'] ##Delete key and value pair
print(student)
```

```python
##Dictionary methods

keys = student.keys() ## get all values 
print(keys)
values = student.values() ## get all values 
print(values)

items = student.items() ## get all key value pair
print(items)
```

```python
## Shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Tarikul2"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy()
print(student_copy1)
print(student)
```

```python
student["name"] = "Tarikul3"
print(student_copy1)
print(student)
```

```python
## Iterating over dictionary 
## You can use loops to iterate over dictionaries, keys, values, or items
## Iterating over keys 

for keys in student.keys():
    print(keys)
```

```python
## Iterate over values 
for value in student.values():
    print(value)
```

```python
## Iterating over key value pair
for key, value in student.items():
    print(f"{key}:{value}")
```
