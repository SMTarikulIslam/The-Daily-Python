```python
## Introduction to Tuples 
# Explanation:
# Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different. 
```

```python
#Creating a tuple

empty_tuple =()
print(empty_tuple)
print(type(empty_tuple))
```

```python
list = list()
print(type(list))
tp1 = tuple()
print(type(tp1))
```

```python
numbers = tuple([1,2,3,4,5,6])
numbers
```

```python
##Immutable nature of Tuples 
## Tuples are immutable their elements cannot be changed once assigned.

list = [1,2,3,4,5]
print(list)

list[1] = "Tarikul"
print(list)
```

```python
## Tuple method
print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing and unpacking tuple 
# Packing
packed_tuple = 1,"Hello", 3.14
print(packed_tuple)
```

```python
# Unpacking tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle, last =numbers
print(first)
print(middle)
print(last)
```

```python
##Nested Tuples 
##Nested List
list1 =[[1,2,3,4], [6,7,8,9], [1,"Hello", 3.14, "c"]]
list1[0]
list1[0][0:3]
```

```python
nested_tuples = ((1,2,3), ("a","b","c"), (True, False))
## Access the elements inside a tuple
print(nested_tuples[0])
print(nested_tuples[1][2])
```

```python
## Iterating over nested tuples
for sub_tuple in nested_tuples:
    for item in sub_tuple:
        print(item, end =" ")
    print()
```
